```python
import pandas as pd

# Read the two CSVs
df1 = pd.read_csv('./data/raw/EMOMUSIC - clips_45seconds/songs_info.csv')
df2 = pd.read_csv('./data/raw/EMOMUSIC - clips_45seconds/static_annotations.csv')

# Merge the DataFrames on the 'song_id' column
df = pd.merge(df1, df2, on='song_id', how='outer')  # Use 'inner' if you want only matching rows

# Save the result
df.to_csv('./data/processed/EMOMUSIC_merged_info.csv', index=False)
```
