```python
# Dynamically access klink root and subsequent paths
from pathlib import Path

project_root = Path.cwd().resolve().parent
csv_path = project_root / "data" / "processed"
raw_path = project_root / "data" / "raw"
```

```python
import pandas as pd

# Read the two CSVs
df1 = pd.read_csv(raw_path / 'EMOMUSIC - clips_45seconds/songs_info.csv')
df2 = pd.read_csv(raw_path / 'EMOMUSIC - clips_45seconds/static_annotations.csv')

# Merge the DataFrames on the 'song_id' column
df = pd.merge(df1, df2, on='song_id', how='outer')  # Use 'inner' if you want only matching rows
```

```python
#Check newly created Data Frame
print(df)
```

```python
# Save the result
df.to_csv(csv_path / 'EMOMUSIC_merged_info.csv', index=False)
```
